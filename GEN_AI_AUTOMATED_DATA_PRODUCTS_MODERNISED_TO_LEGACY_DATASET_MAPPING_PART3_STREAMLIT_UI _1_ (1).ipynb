````
import streamlit as st
from snowflake.snowpark.context import get_active_session
import json

# Get the current Snowflake session
session = get_active_session()

st.title("Data Comparison and Mapping Tool")

# Initialize session state variables
if 'prompt' not in st.session_state:
    st.session_state.prompt = ""
if 'ai_response' not in st.session_state:
    st.session_state.ai_response = ""
if 'token_count' not in st.session_state:
    st.session_state.token_count = 0

# Input fields for database, schema, and table names
database_name = st.text_input("Enter Database Name", "SMART_AI_MAPPER")
schema_name = st.text_input("Enter Schema Name", "SMART_AI_MAPPER_TOOL")
modern_table_name = st.text_input("Enter Modernised Table Name", "Deposit_Accounts_Modernised")
legacy_table_name = st.text_input("Enter Legacy Table Name", "Deposit_Accounts_Legacy")
record_limit = st.number_input("Enter Sample Record Limit", min_value=1, value=5)

# Model selection
model = st.selectbox("Select AI Model", [
    "snowflake-arctic", "mistral-large", "reka-flash", "reka-core", "mixtral-8x7b",
    "jamba-instruct", "llama2-70b-chat", "llama3-8b", "llama3-70b", "llama3.1-8b",
    "llama3.1-70b", "llama3.1-405b", "mistral-7b", "gemma-7b"
])

# Options for prompt generation
use_vector_embeddings = st.checkbox("Use Vector Embeddings")
use_sample_data = st.checkbox("Use Sample Data")

def fetch_available_embeddings(modern_table, legacy_table):
    query = f"""
    SELECT VECTOR_MAPPING_FOR_TABLE_NAME 
    FROM SMART_AI_MAPPER.SMART_AI_MAPPER_TOOL.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
    WHERE UPPER(VECTOR_MAPPING_FOR_TABLE_NAME) LIKE '%{modern_table.upper()}%' 
    AND UPPER(VECTOR_MAPPING_FOR_TABLE_NAME) LIKE '%{legacy_table.upper()}%'
    GROUP BY 1
    """
    result = session.sql(query).collect()
    return [row['VECTOR_MAPPING_FOR_TABLE_NAME'] for row in result]

# Fetch available embeddings
available_embeddings = fetch_available_embeddings(modern_table_name, legacy_table_name)

# Dropdown for selecting embedding
selected_embedding = None
if available_embeddings:
    selected_embedding = st.selectbox("Select Vector Embedding", available_embeddings)
else:
    st.warning("No matching vector embeddings found for the given table names.")

def generate_prompt():
    combined_data = {}
    prompt_text = "Given the provided data, compare the fields between the Modernised and Legacy tables. "
    
    if use_vector_embeddings and selected_embedding:
        # Fetch vector embeddings and mappings
        vector_mappings = session.sql(f"""
        SELECT ARRAY_AGG(
            OBJECT_CONSTRUCT(*)) AS JSON_DATA FROM
            (
        SELECT VECTOR_MAPPING_FOR_TABLE_NAME,        
         LISTAGG(CONCAT(MODERNISED_TABLE_FIELD_NAME,'-',TOP_SIMILARITIES_LEGACY_FIELDS,',')) within group 
            (ORDER BY MODERNISED_TABLE_FIELD_NAME ) AS Modern_table_field_to_legacy_table_mappings
        FROM  SMART_AI_MAPPER.SMART_AI_MAPPER_TOOL.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
        WHERE VECTOR_MAPPING_FOR_TABLE_NAME = '{selected_embedding}'
        GROUP BY 1)
        """).collect()[0]['JSON_DATA']
        combined_data["vector_mappings"] = vector_mappings
        prompt_text += "Use the vector embeddings to help map fields between the Modernised table and the Legacy table. "
    
    if use_sample_data:
        # Call the stored procedure for modern table
        modern_result = session.call('fetch_and_process_table_data', 
                                     database_name, 
                                     schema_name, 
                                     modern_table_name, 
                                     record_limit)
        
        # Call the stored procedure for legacy table
        legacy_result = session.call('fetch_and_process_table_data', 
                                     database_name, 
                                     schema_name, 
                                     legacy_table_name, 
                                     record_limit)
        
        combined_data["modern_data"] = modern_result
        combined_data["legacy_data"] = legacy_result
        prompt_text += "Analyze the sample data to identify discrepancies and provide a mapping between the two tables. Highlight any differences in field names or data values. "
    
    prompt_text += """
    For fields with mismatches, suggest corrections and as per mapping identified, provide a Snowflake compatiable SQL query to transform data from Modernised table to align with the Legacy Table. 
    Ensure that all data matches correctly between Modernised and Legacy, Report if any discrepancies and apply if transformations required in the snowflake 
    SQL generated.
    If there are fields in one table that do not have direct matches in the other table, note these discrepancies and indicate how to handle them.
    
    Data: {data}
    """
    
    st.session_state.prompt = prompt_text.format(data=json.dumps(combined_data))

def get_ai_response(options):
    try:
        # Call the CORTEX.COMPLETE function directly
        result = session.sql(f"""
        SELECT SNOWFLAKE.CORTEX.COMPLETE(
            '{model}',
            ARRAY_CONSTRUCT(OBJECT_CONSTRUCT('role', 'user', 'content', '{st.session_state.prompt.replace("'", "''")}')),
            OBJECT_CONSTRUCT('temperature', {options['temperature']}, 'max_tokens', {options['max_tokens']}, 'top_p', {options['top_p']})
        ) AS response
        """).collect()
        
        if result and len(result) > 0:
            response_json = json.loads(result[0]['RESPONSE'])
            st.session_state.ai_response = response_json
        else:
            st.error("No response received from the AI model.")
    except Exception as e:
        st.error(f"An error occurred: {str(e)}")
        st.session_state.ai_response = "Error: Unable to get AI response."

if st.button("Generate Prompt"):
    generate_prompt()
    
    # Calculate token count
    try:
        token_count_result = session.sql(f"""
        SELECT SNOWFLAKE.CORTEX.COUNT_TOKENS(
            '{model}',
            '{st.session_state.prompt.replace("'", "''")}'
        ) AS token_count
        """).collect()
        
        if token_count_result and len(token_count_result) > 0:
            st.session_state.token_count = token_count_result[0]['TOKEN_COUNT']
        else:
            st.warning("Unable to calculate token count.")
    except Exception as e:
        st.warning(f"Error calculating token count: {str(e)}")

if st.session_state.prompt:
    st.subheader("Generated Prompt:")
    st.text_area("Prompt", st.session_state.prompt, height=300)
    st.write(f"Token Count: {st.session_state.token_count}")
    
    st.subheader("Model Parameters")
    temperature = st.slider("Temperature", 0.0, 1.0, 0.7, 0.1)
    max_tokens = st.number_input("Max Tokens", 1, 4096, min(st.session_state.token_count, 4096))
    top_p = st.slider("Top P", 0.0, 1.0, 1.0, 0.1)
    
    if st.button("Get AI Response"):
        options = {
            "temperature": temperature,
            "max_tokens": max_tokens,
            "top_p": top_p,
        }
        get_ai_response(options)

if st.session_state.ai_response:
    st.subheader("AI Model Response:")
    if isinstance(st.session_state.ai_response, dict):
        if 'choices' in st.session_state.ai_response and len(st.session_state.ai_response['choices']) > 0:
            message = st.session_state.ai_response['choices'][0].get('messages', '')
            st.markdown(message)
        
        if 'usage' in st.session_state.ai_response:
            usage = st.session_state.ai_response['usage']
            st.write("Token Usage:")
            st.write(f"- Completion tokens: {usage.get('completion_tokens', 'N/A')}")
            st.write(f"- Prompt tokens: {usage.get('prompt_tokens', 'N/A')}")
            st.write(f"- Total tokens: {usage.get('total_tokens', 'N/A')}")
        
        if 'model' in st.session_state.ai_response:
            st.write(f"Model used: {st.session_state.ai_response['model']}")
    else:
        st.write(st.session_state.ai_response)
````

````
import streamlit as st
from snowflake.snowpark.context import get_active_session
import json
import datetime
import decimal

# Get the current Snowflake session
session = get_active_session()

def json_serializer(obj):
    """Custom JSON serializer for objects not serializable by default json code"""
    if isinstance(obj, (datetime.datetime, datetime.date)):
        return obj.isoformat()
    elif isinstance(obj, datetime.time):
        return obj.isoformat()
    elif isinstance(obj, decimal.Decimal):
        return float(obj)
    elif hasattr(obj, '__dict__'):
        return str(obj)
    else:
        return str(obj)

def safe_json_dumps(data):
    """Safely serialize data to JSON with custom serializer"""
    try:
        return json.dumps(data, separators=(',', ':'), default=json_serializer)
    except Exception as e:
        st.error(f"JSON serialization error: {str(e)}")
        # Fallback: convert problematic objects to strings
        return json.dumps(str(data), separators=(',', ':'))

st.title("Data Comparison and Mapping Tool")

# Initialize session state variables
if 'prompt' not in st.session_state:
    st.session_state.prompt = ""
if 'ai_response' not in st.session_state:
    st.session_state.ai_response = ""
if 'token_count' not in st.session_state:
    st.session_state.token_count = 0

# Input fields for database, schema, and table names
database_name = st.text_input("Enter Database Name", "SMART_AI_MAPPER")
schema_name = st.text_input("Enter Schema Name", "SMART_AI_MAPPER_TOOL")
modern_table_name = st.text_input("Enter Modernised Table Name", "Deposit_Accounts_Modernised")
legacy_table_name = st.text_input("Enter Legacy Table Name", "Deposit_Accounts_Legacy")

# Add token optimization controls
st.subheader("Token Optimization")
col1, col2 = st.columns(2)
with col1:
    record_limit = st.number_input("Sample Record Limit", min_value=1, max_value=5, value=2, help="Lower values reduce token usage")
with col2:
    max_fields = st.number_input("Max Fields to Include", min_value=5, max_value=20, value=10, help="Limit fields to reduce token usage")

# Model selection
model = st.selectbox("Select AI Model", [
    "snowflake-arctic", "mistral-large", "reka-flash", "reka-core", "mixtral-8x7b",
    "jamba-instruct", "llama2-70b-chat", "llama3-8b", "llama3-70b", "llama3.1-8b",
    "llama3.1-70b", "llama3.1-405b", "mistral-7b", "gemma-7b"
])

# Options for prompt generation
use_vector_embeddings = st.checkbox("Use Vector Embeddings")
use_sample_data = st.checkbox("Use Sample Data")

def set_session_context(database_name, schema_name):
    """Set the session context to ensure proper database/schema access"""
    try:
        session.sql(f"USE DATABASE {database_name}").collect()
        session.sql(f"USE SCHEMA {database_name}.{schema_name}").collect()
        st.success(f"Session context set to {database_name}.{schema_name}")
        return True
    except Exception as e:
        st.error(f"Failed to set session context: {str(e)}")
        return False

def check_stored_procedure_exists(database_name, schema_name, procedure_name):
    """Check if the stored procedure exists in the current schema"""
    try:
        result = session.sql(f"""
        SHOW PROCEDURES LIKE '{procedure_name}' IN SCHEMA {database_name}.{schema_name}
        """).collect()
        return len(result) > 0
    except Exception as e:
        st.warning(f"Could not check procedure existence: {str(e)}")
        return False

def fetch_available_embeddings(modern_table, legacy_table):
    query = f"""
    SELECT VECTOR_MAPPING_FOR_TABLE_NAME 
    FROM {database_name}.{schema_name}.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
    WHERE UPPER(VECTOR_MAPPING_FOR_TABLE_NAME) LIKE '%{modern_table.upper()}%' 
    AND UPPER(VECTOR_MAPPING_FOR_TABLE_NAME) LIKE '%{legacy_table.upper()}%'
    GROUP BY 1
    """
    try:
        result = session.sql(query).collect()
        return [row['VECTOR_MAPPING_FOR_TABLE_NAME'] for row in result]
    except Exception as e:
        st.error(f"Error fetching embeddings: {str(e)}")
        return []

def fetch_sample_data_direct(database_name, schema_name, table_name, limit):
    """Fallback method to fetch sample data directly if stored procedure fails"""
    try:
        # Get table structure (limited fields)
        structure_query = f"""
        SELECT COLUMN_NAME, DATA_TYPE 
        FROM {database_name}.INFORMATION_SCHEMA.COLUMNS 
        WHERE TABLE_SCHEMA = '{schema_name}' 
        AND TABLE_NAME = '{table_name.upper()}'
        ORDER BY ORDINAL_POSITION
        LIMIT {max_fields}
        """
        
        structure_result = session.sql(structure_query).collect()
        columns = [{"column_name": row['COLUMN_NAME'], "data_type": row['DATA_TYPE']} for row in structure_result]
        
        # Build column list for data query (limit to selected columns)
        column_names = [col['column_name'] for col in columns]
        column_list = ', '.join(column_names)
        
        # Get sample data with limited columns
        data_query = f"SELECT {column_list} FROM {database_name}.{schema_name}.{table_name} LIMIT {limit}"
        data_result = session.sql(data_query).collect()
        
        # Convert to dictionary format (simplified)
        sample_data = []
        for row in data_result:
            row_dict = {}
            for col_name in column_names:
                value = row[col_name] if col_name in row.as_dict() else None
                
                # Handle different data types for JSON serialization
                if value is not None:
                    if isinstance(value, (datetime.datetime, datetime.date)):
                        value = value.isoformat()
                    elif isinstance(value, datetime.time):
                        value = value.isoformat()
                    elif isinstance(value, decimal.Decimal):
                        value = float(value)
                    elif isinstance(value, str) and len(value) > 50:
                        value = value[:47] + "..."
                    elif not isinstance(value, (str, int, float, bool, type(None))):
                        value = str(value)
                
                row_dict[col_name] = value
            sample_data.append(row_dict)
        
        return {
            "table_name": table_name,
            "structure": columns,
            "sample_data": sample_data,
            "record_count": len(sample_data)
        }
        
    except Exception as e:
        st.error(f"Error fetching sample data for {table_name}: {str(e)}")
        return None

# Set session context button
if st.button("Set Session Context"):
    set_session_context(database_name, schema_name)

# Check if stored procedure exists
if st.button("Check Stored Procedure"):
    exists = check_stored_procedure_exists(database_name, schema_name, "FETCH_AND_PROCESS_TABLE_DATA")
    if exists:
        st.success("✅ Stored procedure FETCH_AND_PROCESS_TABLE_DATA exists")
    else:
        st.warning("⚠️ Stored procedure FETCH_AND_PROCESS_TABLE_DATA not found")
        st.info("The app will use direct table queries as fallback")

# Fetch available embeddings
available_embeddings = fetch_available_embeddings(modern_table_name, legacy_table_name)

# Dropdown for selecting embedding
selected_embedding = None
if available_embeddings:
    selected_embedding = st.selectbox("Select Vector Embedding", available_embeddings)
else:
    st.warning("No matching vector embeddings found for the given table names.")

# Add debugging section
st.subheader("🔍 Debug Information")
if st.button("Show Available Vector Mappings"):
    try:
        debug_query = f"""
        SELECT 
            VECTOR_MAPPING_FOR_TABLE_NAME,
            COUNT(*) as field_count,
            LISTAGG(MODERNISED_TABLE_FIELD_NAME, ', ') WITHIN GROUP (ORDER BY MODERNISED_TABLE_FIELD_NAME) as modern_fields
        FROM {database_name}.{schema_name}.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
        GROUP BY VECTOR_MAPPING_FOR_TABLE_NAME
        ORDER BY VECTOR_MAPPING_FOR_TABLE_NAME
        """
        debug_results = session.sql(debug_query).collect()
        
        st.write("**Available Vector Mapping Tables:**")
        for row in debug_results:
            st.write(f"- **{row['VECTOR_MAPPING_FOR_TABLE_NAME']}**: {row['FIELD_COUNT']} fields")
            st.write(f"  Fields: {row['MODERN_FIELDS']}")
            st.write("")
            
    except Exception as e:
        st.error(f"Debug query failed: {str(e)}")

if st.button("Preview Selected Vector Mapping"):
    if selected_embedding:
        try:
            preview_query = f"""
            SELECT 
                MODERNISED_TABLE_FIELD_NAME,
                TOP_SIMILARITIES_LEGACY_FIELDS
            FROM {database_name}.{schema_name}.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
            WHERE VECTOR_MAPPING_FOR_TABLE_NAME = '{selected_embedding}'
            ORDER BY MODERNISED_TABLE_FIELD_NAME
            LIMIT 5
            """
            preview_results = session.sql(preview_query).collect()
            
            st.write(f"**Preview of {selected_embedding}:**")
            for row in preview_results:
                st.write(f"**{row['MODERNISED_TABLE_FIELD_NAME']}** → {row['TOP_SIMILARITIES_LEGACY_FIELDS']}")
                
        except Exception as e:
            st.error(f"Preview query failed: {str(e)}")
    else:
        st.warning("Please select a vector embedding first")

def truncate_data_for_prompt(data, max_items=3):
    """Truncate data to fit within token limits"""
    if isinstance(data, dict):
        if "sample_data" in data and isinstance(data["sample_data"], list):
            # Keep only first few records
            data["sample_data"] = data["sample_data"][:max_items]
        if "structure" in data and isinstance(data["structure"], list):
            # Keep structure but limit to essential info
            data["structure"] = [{"column_name": col.get("column_name", ""), "data_type": col.get("data_type", "")} for col in data["structure"][:10]]
    return data

def generate_prompt():
    # Set session context first
    if not set_session_context(database_name, schema_name):
        st.error("Cannot proceed without proper session context")
        return
    
    combined_data = {}
    prompt_text = "You are helping legacy users access modernized database tables using their familiar legacy field names. "
    
    if use_vector_embeddings and selected_embedding:
        try:
            # Fetch comprehensive vector embeddings and mappings
            vector_mappings_query = f"""
            SELECT 
                MODERNISED_TABLE_FIELD_NAME,
                TOP_SIMILARITIES_LEGACY_FIELDS
            FROM {database_name}.{schema_name}.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
            WHERE VECTOR_MAPPING_FOR_TABLE_NAME = '{selected_embedding}'
            ORDER BY MODERNISED_TABLE_FIELD_NAME
            """
            
            vector_results = session.sql(vector_mappings_query).collect()
            
            # Parse the similarity strings to extract individual mappings
            detailed_mappings = []
            for row in vector_results:
                modern_field = row['MODERNISED_TABLE_FIELD_NAME']
                similarities_str = row['TOP_SIMILARITIES_LEGACY_FIELDS']
                
                # Parse the similarity string (format: "field1-Similarity0.95|field2-Similarity0.85|field3-Similarity0.75|")
                if similarities_str:
                    similarity_parts = similarities_str.strip('|').split('|')
                    legacy_matches = []
                    
                    for part in similarity_parts:
                        if '-Similarity' in part:
                            field_name = part.split('-Similarity')[0]
                            similarity_score = part.split('-Similarity')[1]
                            legacy_matches.append({
                                "field": field_name,
                                "similarity": similarity_score
                            })
                    
                    detailed_mappings.append({
                        "modern_field": modern_field,
                        "legacy_matches": legacy_matches
                    })
            
            combined_data["vector_mappings"] = {
                "mapping_table": selected_embedding,
                "field_mappings": detailed_mappings,
                "total_fields": len(detailed_mappings)
            }
            
            prompt_text += f"Use the vector embeddings showing {len(detailed_mappings)} field mappings to create the legacy-to-modern field mapping. "
            
            # Log the mappings for debugging
            st.info(f"✅ Loaded {len(detailed_mappings)} field mappings from vector embeddings")
            
        except Exception as e:
            st.error(f"Error fetching vector mappings: {str(e)}")
            # Show the actual query for debugging
            st.code(f"""
            Query used:
            SELECT 
                MODERNISED_TABLE_FIELD_NAME,
                TOP_SIMILARITIES_LEGACY_FIELDS
            FROM {database_name}.{schema_name}.TOP_3_SIMILAR_FIELDS_FROM_VEC_EMB
            WHERE VECTOR_MAPPING_FOR_TABLE_NAME = '{selected_embedding}'
            ORDER BY MODERNISED_TABLE_FIELD_NAME
            """)
    
    if use_sample_data:
        try:
            # First try using the stored procedure with fully qualified name
            if check_stored_procedure_exists(database_name, schema_name, "FETCH_AND_PROCESS_TABLE_DATA"):
                try:
                    # Try calling with fully qualified procedure name
                    modern_result = session.call(f'{database_name}.{schema_name}.fetch_and_process_table_data', 
                                                 database_name, 
                                                 schema_name, 
                                                 modern_table_name, 
                                                 min(record_limit, 3))  # Limit to max 3 records
                    
                    legacy_result = session.call(f'{database_name}.{schema_name}.fetch_and_process_table_data', 
                                                 database_name, 
                                                 schema_name, 
                                                 legacy_table_name, 
                                                 min(record_limit, 3))  # Limit to max 3 records
                    
                    combined_data["modern_data"] = truncate_data_for_prompt(modern_result)
                    combined_data["legacy_data"] = truncate_data_for_prompt(legacy_result)
                    st.success("✅ Successfully fetched data using stored procedure")
                    
                except Exception as proc_error:
                    st.warning(f"Stored procedure failed: {str(proc_error)}")
                    st.info("Falling back to direct table queries...")
                    raise proc_error
            else:
                raise Exception("Stored procedure not found")
                
        except Exception as e:
            # Fallback to direct queries
            st.info("Using direct table queries...")
            modern_result = fetch_sample_data_direct(database_name, schema_name, modern_table_name, min(record_limit, 3))
            legacy_result = fetch_sample_data_direct(database_name, schema_name, legacy_table_name, min(record_limit, 3))
            
            if modern_result and legacy_result:
                combined_data["modern_data"] = truncate_data_for_prompt(modern_result)
                combined_data["legacy_data"] = truncate_data_for_prompt(legacy_result)
                st.success("✅ Successfully fetched data using direct queries")
            else:
                st.error("Failed to fetch sample data using both methods")
                return
        
        prompt_text += "Analyze the sample data from both tables to understand field differences and data formats. "
    
    # Add table context to the prompt
    combined_data["table_context"] = {
        "modern_table": f"{database_name}.{schema_name}.{modern_table_name}",
        "legacy_table": f"{database_name}.{schema_name}.{legacy_table_name}",
        "purpose": "Create SQL query for legacy users to access modernized table"
    }
    
    prompt_text += f"""
    Your task is to help legacy users query the modernized table using their familiar legacy field names.
    
    Please provide:
    1. A field mapping table showing: Legacy_Field_Name → Modern_Field_Name
    2. A Snowflake SQL query that selects from the modernized table but uses legacy field names as aliases
    3. Note any fields that don't have direct matches and suggest handling approaches
    4. Include any necessary data transformations (format changes, calculations, etc.)
    
    Example output format:
    
    **Field Mapping:**
    - Cust_ID → Customer_ID
    - Balance → Account_Balance
    - Acct_Type → Account_Type_Code
    
    **SQL Query for Legacy Users:**
    ```sql
    SELECT 
        Customer_ID AS Cust_ID,
        Account_Balance AS Balance,
        Account_Type_Code AS Acct_Type,
        -- Add transformations if needed
        CASE WHEN Status = 'ACTIVE' THEN 'A' ELSE 'I' END AS Status_Code
    FROM {database_name}.{schema_name}.{modern_table_name}
    WHERE condition_if_needed;
    ```
    
    **Notes:**
    - List any unmapped fields
    - Mention required transformations
    - Suggest handling for missing legacy fields
    
    Data: {{data}}
    """
    
    # Create compact JSON without indentation to save tokens
    data_json = safe_json_dumps(combined_data)
    st.session_state.prompt = prompt_text.format(data=data_json)

def get_ai_response(options):
    try:
        # Escape single quotes properly
        escaped_prompt = st.session_state.prompt.replace("'", "''")
        
        # Call the CORTEX.COMPLETE function directly
        result = session.sql(f"""
        SELECT SNOWFLAKE.CORTEX.COMPLETE(
            '{model}',
            ARRAY_CONSTRUCT(OBJECT_CONSTRUCT('role', 'user', 'content', '{escaped_prompt}')),
            OBJECT_CONSTRUCT('temperature', {options['temperature']}, 'max_tokens', {options['max_tokens']}, 'top_p', {options['top_p']})
        ) AS response
        """).collect()
        
        if result and len(result) > 0:
            response_json = json.loads(result[0]['RESPONSE'])
            st.session_state.ai_response = response_json
        else:
            st.error("No response received from the AI model.")
    except Exception as e:
        st.error(f"An error occurred: {str(e)}")
        st.session_state.ai_response = "Error: Unable to get AI response."

if st.button("Generate Prompt"):
    generate_prompt()
    
    # Calculate token count
    if st.session_state.prompt:
        try:
            escaped_prompt = st.session_state.prompt.replace("'", "''")
            token_count_result = session.sql(f"""
            SELECT SNOWFLAKE.CORTEX.COUNT_TOKENS(
                '{model}',
                '{escaped_prompt}'
            ) AS token_count
            """).collect()
            
            if token_count_result and len(token_count_result) > 0:
                st.session_state.token_count = token_count_result[0]['TOKEN_COUNT']
            else:
                st.warning("Unable to calculate token count.")
        except Exception as e:
            st.warning(f"Error calculating token count: {str(e)}")

if st.session_state.prompt:
    st.subheader("Generated Prompt:")
    st.text_area("Prompt", st.session_state.prompt, height=300)
    st.write(f"Token Count: {st.session_state.token_count}")
    
    st.subheader("Model Parameters")
    temperature = st.slider("Temperature", 0.0, 1.0, 0.7, 0.1)
    # Calculate safe max tokens based on prompt length
    if st.session_state.token_count > 0:
        # Leave buffer for response (at least 500 tokens)
        safe_max_tokens = min(4096 - st.session_state.token_count - 100, 3000)
        if safe_max_tokens < 100:
            safe_max_tokens = 100
    else:
        safe_max_tokens = 1000
    
    max_tokens = st.number_input("Max Tokens", 1, 3000, safe_max_tokens)
    
    # Show warning if prompt is too long
    if st.session_state.token_count > 3500:
        st.warning(f"⚠️ Prompt is very long ({st.session_state.token_count} tokens). Consider reducing sample size or disabling some options.")
    elif st.session_state.token_count > 3000:
        st.info(f"ℹ️ Large prompt ({st.session_state.token_count} tokens). Response will be limited.")
    top_p = st.slider("Top P", 0.0, 1.0, 1.0, 0.1)
    
    if st.button("Get AI Response"):
        options = {
            "temperature": temperature,
            "max_tokens": max_tokens,
            "top_p": top_p,
        }
        get_ai_response(options)

if st.session_state.ai_response:
    st.subheader("AI Model Response:")
    if isinstance(st.session_state.ai_response, dict):
        if 'choices' in st.session_state.ai_response and len(st.session_state.ai_response['choices']) > 0:
            message = st.session_state.ai_response['choices'][0].get('messages', '')
            st.markdown(message)
        
        if 'usage' in st.session_state.ai_response:
            usage = st.session_state.ai_response['usage']
            st.write("Token Usage:")
            st.write(f"- Completion tokens: {usage.get('completion_tokens', 'N/A')}")
            st.write(f"- Prompt tokens: {usage.get('prompt_tokens', 'N/A')}")
            st.write(f"- Total tokens: {usage.get('total_tokens', 'N/A')}")
        
        if 'model' in st.session_state.ai_response:
            st.write(f"Model used: {st.session_state.ai_response['model']}")
    else:
        st.write(st.session_state.ai_response)
````
